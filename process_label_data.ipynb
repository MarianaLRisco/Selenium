```python
import pandas as pd
import os
```

```python
# Cargar el archivo CSV
df = pd.read_csv("dataset\dataset.csv", header=None, names=["class_label", "image_path", "post_text"])

df["image_path"] = df["image_path"].apply(lambda x: os.path.join("dataset", os.path.basename(x)).replace("\\", "/"))

label_translation = {
    "enchente": ["flood"],
    "deslizamento_de_terra":["landslide"]
}

df["labels"] = df["class_label"].map(label_translation)
df = df.drop(columns=["class_label"])

# Guardar el nuevo CSV
df.to_csv("dataset_multilabel.csv", index=False)
```
