```python
data = pd.read_csv("data/disasterDataset/datasetDisaster/dataset_curate1.csv")  
# data = fix_paths(data)

corrupt_images = []

for idx, row in data.iterrows():
    img_path = row['image_path']
    img_path = os.path.join("data/disasterDataset/", img_path).replace("\\", "/")
    if not os.path.exists(img_path):
        print(f"No existe: {img_path}")
    try:
        with Image.open(img_path) as img:
            img.verify()  
    except Exception as e:
        print(f"Corrupt image: {img_path} -> {e}")
        corrupt_images.append(img_path)

print(f"\nTotal de imágenes corruptas: {len(corrupt_images)}")
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.vgg16 import preprocess_input
from PIL import ImageFile
import numpy as np

def get_input(path):
    ImageFile.LOAD_TRUNCATED_IMAGES = True
    img = load_img(path, target_size=(150,150))
    return img


# Takes in image and preprocess it
def process_input(img):
    img_data = img_to_array(img)        
    img_data = np.expand_dims(img_data, axis=0) 
    img_data = preprocess_input(img_data)     
    return img_data
```

```python
dt = pd.read_csv("data/damage_dataset/dataset.csv")
dt.head()

data_path = "data/damage_dataset"
```

```python
dt["image_path"] = dt["image_path"].str.replace("^dataset/", "", regex=True)
dt.head()
```

```python
def corrupt_images(path, data):
    corrupt_images = []

    for idx, row in data.iterrows():
        img_path = row['image_path']
        img_path = os.path.join(path, img_path).replace("\\", "/")
        if not os.path.exists(img_path):
            print(f"No existe: {img_path}")
        try:
            with Image.open(img_path) as img:
                img.verify()  
        except Exception as e:
            print(f"Corrupt image: {img_path} -> {e}")
            corrupt_images.append(img_path)

    print(f"\nTotal de imágenes corruptas: {len(corrupt_images)}")

corrupt_images(data_path,dt)
```
