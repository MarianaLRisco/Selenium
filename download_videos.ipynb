```python
# Buscar por hashtag
hashtag = "robo"
hashtag1 = "camaras"
hashtag2 = "Trujillo"
search_url = f"https://www.facebook.com/search/top?q={hashtag}%20{hashtag1}%20{hashtag2}"
driver.get(search_url)
time.sleep(5)

# Scroll automático y descarga de videos
videos = set()
scroll_pause_time = 3
max_videos = 100
output_dir = "Dataset/robo"

# Configurar yt_dlp
ydl_opts = {
    'outtmpl': os.path.join(output_dir, '%(title).50s.%(ext)s'),
    'quiet': False,
    'format': 'best',
}

with YoutubeDL(ydl_opts) as ydl:
    while len(videos) < max_videos:
        driver.find_element(By.TAG_NAME, "body").send_keys(Keys.END)
        time.sleep(random.uniform(2.5, 4.5))

        posts = driver.find_elements(By.XPATH, '//a[contains(@href, "/videos/")]')
        for post in posts:
            href = post.get_attribute("href")
            if href and "/videos/" in href and href not in videos:
                videos.add(href)
                print(f"🎥 Nuevo video encontrado: {href}")
                try:
                    print("⬇️ Descargando...")
                    ydl.download([href])
                except Exception as e:
                    print(f"❌ Error al descargar {href}: {e}")

        print(f"Total descargados: {len(videos)}")
```
